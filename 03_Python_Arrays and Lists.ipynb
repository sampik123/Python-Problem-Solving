#### Q5. Reverse List

```python
def Reverse_List(li):
    length = len(li)
    for i in range(length//2):
        li[i], li[-i-1] = li[-i-1], li[i]
    return li


li = [ 1,2,3,4,5]
Reverse_List(li)
print(li)
```

#### Q8. Swap Alternate -- given an array. you need to swap every pair of alternate elements in the array.

```python
def Swap_Alternate(li):
    length = len(li)
    for i in range(0, length-1 , 2):
        li[i], li[i+1] = li[i+1], li[i]
        
    return li

li = [1,2,3,4,5,6]
Swap_Alternate(li)
print(li)
```

#### Q9. Find Unique -- all numbers are present twice except one number, find that.

```python
def Find_Unique(arr,n):
    for i in range(n):
        count = 0
        for j in range(n):
            if arr[i] == arr[j]:
                count = count + 1
                
        if count == 1:
            return arr[i]
        
arr = [1,2,6,1,2,5,6]
value = Find_Unique(arr,7)
print(value)
```

#### Q12. Pair Sum -- you have been given an integer array/list and a number X. find and return the total number of pairs in the array which sum to X

```python
def Pair_Sum(arr,n,x):
    count = 0
    for i in range(n):
       
        for j in range(i+1, n):
            if arr[i] + arr[j] == x:
                count = count + 1
                
    return count
    
arr = [6,3,4,6,1,7,2,6]
n = 8
x = 7
var = Pair_Sum(arr,n,x)
print(var)
```

#### Q13. Triplet Sum -- you have been given a random integer array/list and a number X. find and return the number of triplets in the array which sum to X.

```python
def triplet_sum(arr,n,X):
    count = 0
    for i in range(n-2):
        for j in range(i+1,n-1):
            for k in range(j+1,n):
                if arr[i] + arr[j] + arr[k] == X:
                    count = count + 1
                    
    return count

arr = [2,1,4,5,7,3,9,6]
n = 6
X = 10
TripletSum = triplet_sum(arr,n,X)
print(TripletSum)
```

#### Q14. Sort 0 1 -- you have given an integer array of size N that contains only integers 0 and 1 . write a function to sort this array. think of a solution which scans the array only once and dont require use of an extra array

```python
# using two pointers algorithm
def SortZerosAndOne(arr,n):
    s=0
    e=n-1
    while s <= e:
        if arr[s] == 0:
            s = s+1
        elif arr[e] == 1:
            e = e-1
        else:
            arr[s], arr[e] = arr[e], arr[s]
            s = s+1
            e = e-1
            
arr = [1,1,0,0,1,1]
n = 6
SortZerosAndOne(arr,n)
print(arr)
```

#### Q17. Move all the negative elements to one side of the array.

```python
# using two pointers algorithm
def move_negative(arr,n):
    s = 0
    e = n-1
    while s<=e:
        if arr[s] < 0:
            s = s + 1
        elif arr[e] >0:
            e = e - 1
        else:
            arr[s], arr[e] = arr[e], arr[s]
            s = s + 1
            e = e - 1
    return arr        
            
            
arr = [ 2,3,-2,-1,3,-5,-6,5]
n = 8
store = move_negative(arr,n)
print(store)
```

#### Q18. Find the Union of the two sorted arrays.

```python
def union(arr1, arr2):
    length1 = len(arr1)
    length2 = len(arr2)
    li = set()
    for i in range(length1):
        if arr1[i] not in li:
            li.add(arr1[i])
            
    for j in range(length2):
        if arr2[j] not in li:
            li.add(arr2[j])
    return li

arr1 = [1,3,6,7,8]
arr2 = [2,4,5,8,9]
store = union(arr1,arr2)
print(store)
```

#### Q19. Rotates D elements towards Right.

```python
def Rotate_Right(arr, n, d):
    for j in range(d):
        temp = arr[n-1]
        for i in range(n):
            arr[n-i-1] = arr[n-i-2]
        arr[0] = temp
    return arr     
        
arr = [1,2,3,4,5]
n = 5
d = 2
print(Rotate_Right(arr, n , d))       
```

#### Q20. find Largest sum contiguous Subarray [V. IMP]......Given an array Arr[] of N integers. Find the contiguous sub-array(containing at least one number) which has the maximum sum and return its sum.

```python
# brute force approach
def Max_Sum_Subarray(arr):
    n = len(arr)
    max_sum = float('-inf')
    start , end = 0 , 0
    
    for i in range(n):
        current_sum = 0
        for j in range(i,n):
            current_sum = current_sum + arr[j]
            if current_sum > max_sum:
                max_sum = current_sum
                start, end = i , j
    return arr[start:end+1], max_sum

arr = [-5,4,6,-3,4,-1]
store = Max_Sum_Subarray(arr)
print(store)          
```

#### Q21. Given three arrays sorted in increasing order. Find the elements that are common in all three arrays.

```python
def find_common_elements(arr1, arr2, arr3):
    common_elements = []
    i = j = k = 0

    while i < len(arr1) and j < len(arr2) and k < len(arr3):
        if arr1[i] == arr2[j] == arr3[k]:
            common_elements.append(arr1[i])
            i = i + 1
            j = j + 1
            k = k + 1
        elif arr1[i] < arr2[j]:
            i = i + 1
        elif arr2[j] < arr3[k]:
            j = j + 1
        else:
            k = k + 1
    return common_elements

arr1 = [1,5,10,11,20,80,90]
arr2 = [4,5,6,20,21,25,80,90]
arr3 = [5,10,12,15,20,25,80,90,92,95,96,100]
store = find_common_elements(arr1,arr2,arr3)
print(store)
```

#### Q22. Given an array of positive and negative numbers. Find if there is a subarray (of size at-least one) with 0 sum.

```python
def Zero_Sum_Subarray(arr):
    n = len(arr)
    start, end = -1, -1  # Initialize start and end to -1 by default

    for i in range(n):
        sum = 0
        for j in range(i, n):
            sum += arr[j]
            if sum == 0:
                start, end = i, j

    if start == -1 or end == -1:
        return []  # Return an empty list if no zero-sum subarray is found
    else:
        return arr[start:end + 1]

arr = [-2, 3, 5, -5, 6]
store = Zero_Sum_Subarray(arr)
print(store)
```

#### Q23. Find the first and last positions of an element in a sorted array.

```python
def findFirstAndLast(arr, n, x):
    first = -1
    last = -1
    for i in range(n):
        if x != arr[i]:
            continue
        if first == -1:
            first = i
        last = i
        
    if first != -1:
        print(f"first occurence is{first} and last occurence is {last}")
    else:
        print('not found')
  

arr = [1, 2, 2, 2, 2, 3, 4, 7, 8, 8]
n = len(arr)
x = 2
findFirstAndLast(arr, n, x)
```

```python
#Optimized approach
def find_first_last_pos(arr, n, target):
    # Initialize variables to store the first and last positions
    first_pos = -1
    last_pos = -1
    
    # Binary search to find the first occurrence of the target element
    left, right = 0, n - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            first_pos = mid
            right = mid - 1  # Continue searching towards the left
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    # Binary search to find the last occurrence of the target element
    left, right = 0, n - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            last_pos = mid
            left = mid + 1  # Continue searching towards the right
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    
    return first_pos, last_pos

# Example usage:
arr = [1, 3, 5, 5, 5, 5, 5, 123, 125]
n = len(arr)
target = 5
first, last = find_first_last_pos(arr, n, target)
print("First occurrence:", first)
print("Last occurrence:", last)
```

#### Q28. Running sum of 1D array. Input : [1,2,3,4] and output : [1,3,6,10]

```python
def running_sum(arr, n):
    li = []
    sum = 0
    for i in range(n):
        sum = sum + arr[i]
        li.append(sum)
    return li

arr = [1,2,3,4]
n = 4
store = running_sum(arr, n)
print(store)
```

#### Q29. Find Pivot Index:    Given an array of integers nums, calculate the pivot index of this array. The pivot index is the index where the sum of all the numbers strictly to the left of the index is equal to the sum of all the numbers strictly to the index's right. If the index is on the left edge of the array, then the left sum is 0 because there are no elements to the left. This also applies to the right edge of the array. Return the leftmost pivot index. If no such index exists, return -1.

```python
def pivotIndex(nums):
    n = len(nums)
    
    # Calculate the total sum of the array
    total_sum = sum(nums)
    
    # Initialize left_sum to 0
    left_sum = 0
    
    # Iterate through the array
    for i in range(n):
        # Check if the left sum is equal to the right sum
        if left_sum == total_sum - left_sum - nums[i]:
            return i
        
        # Update left_sum for the next iteration
        left_sum += nums[i]
    
    # If no pivot index is found, return -1
    return -1

# Example usage:
nums = [1, 7, 3, 6, 5, 6]
result = pivotIndex(nums)
print(result)
```
